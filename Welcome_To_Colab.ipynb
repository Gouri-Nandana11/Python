```
import pandas as pd
import numpy as np

# Read dataset
df = pd.read_csv('/content/preprocessed_resume (2).csv')

# Remove hidden spaces from column names
df.columns = df.columns.str.strip()

# Remove hidden BOM character
df.columns = df.columns.str.replace('\ufeff', '', regex=False)

# Replace all NaN values with empty string
df = df.replace(np.nan, "", regex=True)

# Remove duplicate rows
df = df.drop_duplicates()

# Remove leading/trailing spaces from text columns
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].astype(str).str.strip()

# Verify
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Duplicates:", df.duplicated().sum())
print("Null Values:", df.isnull().sum().sum())

# Save cleaned dataset
df.to_csv("cleaned_resume.csv", index=False)
```

```
import pandas as pd
import ast

# Load cleaned dataset
df = pd.read_csv("cleaned_resume.csv", keep_default_na=False)

# Columns that should become Python lists
list_columns = [
    "skills",
    "degree_names",
    "educational_institution_name",
    "professional_company_names",
    "positions",
    "languages",
    "certification_skills",
    "related_skils_in_job"
]

def convert_to_list(value):
    if value == "":
        return []
    try:
        return ast.literal_eval(value)
    except:
        return []

# Convert columns
for col in list_columns:
    df[col] = df[col].apply(convert_to_list)

print("Conversion completed successfully!")
```

```
import pandas as pd
import re
import string
import nltk
import spacy

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
```

```
nltk.download("punkt")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")
```

```
nlp = spacy.load("en_core_web_sm")

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

print("Everything loaded successfully!")
```

```
df = pd.read_pickle("/content/structured_resume.pkl")
```

```
text_columns = [
    "career_objective",
    "educationaL_requirements",
    "experiencere_requirement",
    "age_requirement"
]

def preprocess_text(text):

    if pd.isna(text) or text == "":
        return ""

    text = str(text).lower()

    text = re.sub(r"[^\w\s]", " ", text)
    text = re.sub(r"\d+", " ", text)

    words = word_tokenize(text)

    words = [w for w in words if w not in stop_words]

    words = [lemmatizer.lemmatize(w) for w in words]

    return " ".join(words)
```

```
# Apply NLP preprocessing to text columns

text_columns = [
    "career_objective",
    "educationaL_requirements",
    "experiencere_requirement",
    "responsibilities",
    "responsibilities.1",
    "skills_required"
]

for col in text_columns:
    df[col] = df[col].apply(preprocess_text)


# Normalize list columns (convert all text to lowercase)

list_columns = [
    "skills",
    "degree_names",
    "educational_institution_name",
    "professional_company_names",
    "positions",
    "languages",
    "certification_skills"
]

for col in list_columns:
    df[col] = df[col].apply(
        lambda lst: [str(item).lower().strip() for item in lst]
    )


# Flatten related_skils_in_job if it contains nested lists

def flatten(lst):

    # If the entire value is None or NaN
    if pd.isna(lst):
        return []

    # If it is not a list
    if not isinstance(lst, list):
        return []

    flat = []

    for item in lst:

        # Skip None values
        if item is None:
            continue

        # If nested list
        if isinstance(item, list):
            flat.extend(
                [str(x).lower().strip() for x in item if x is not None]
            )

        else:
            flat.append(str(item).lower().strip())

    return flat

print("✅ NLP preprocessing completed successfully!")
```
